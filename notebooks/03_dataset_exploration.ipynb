```python
import os

cloud_path = r"D:\CloudRemoval_Project\datasets\RICE2\RICE2\cloud"
label_path = r"D:\CloudRemoval_Project\datasets\RICE2\RICE2\label"
mask_path = r"D:\CloudRemoval_Project\datasets\RICE2\RICE2\mask"

print("Cloud Images :", len(os.listdir(cloud_path)))
print("Label Images :", len(os.listdir(label_path)))
print("Mask Images  :", len(os.listdir(mask_path)))
```

```python
import cv2
import matplotlib.pyplot as plt
import os

cloud_img = cv2.imread(
    os.path.join(cloud_path, "0.png")
)

cloud_img = cv2.cvtColor(
    cloud_img,
    cv2.COLOR_BGR2RGB
)

plt.imshow(cloud_img)
plt.title("Cloud Image")
plt.axis("off")
plt.show()
```

```python
label_img = cv2.imread(
    os.path.join(label_path, "0.png")
)

label_img = cv2.cvtColor(
    label_img,
    cv2.COLOR_BGR2RGB
)

plt.imshow(label_img)
plt.title("Label Image")
plt.axis("off")
plt.show()
```

```python
mask_img = cv2.imread(
    os.path.join(mask_path, "0.png"),
    cv2.IMREAD_GRAYSCALE
)

plt.imshow(mask_img, cmap="gray")
plt.title("Mask")
plt.axis("off")
plt.show()
```
