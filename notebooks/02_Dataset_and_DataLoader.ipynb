# from torch.utils.data import Dataset
Why?

Because Dataset is already written by PyTorch engineers.

Inside PyTorch:

torch
 └── utils
      └── data
            └── Dataset

We import it so we can inherit from it.

PyTorch needs a standardized format to talk to your data. It cannot guess how you organized your files on your hard drive.

Here are the primary reasons PyTorch requires this Dataset structure:

Reason A: It Standardizes Data Access
Every machine learning project uses different data types—some use CSV files, some use audio tracks, and others use satellite images.

PyTorch engineers created the Dataset class as a unified interface. No matter what kind of data you are using, PyTorch expects to interact with it using square brackets (dataset[0], dataset[1]). By inheriting from Dataset, you guarantee your custom data structures will seamlessly integrate with the rest of PyTorch's ecosystem.

```python
from torch.utils.data import Dataset
class MyDataset(Dataset):
    def __init__(self):
        self.data = [1,2,3,4,5]
 

    def __len__(self):
        return len(self.data)

    def __getitem__(self,index):
        return self.data[index]

dataset = MyDataset()

print(len(dataset))
```

```python
# 1. First, import DataLoader from PyTorch's library
from torch.utils.data import DataLoader

# 2. Now create the DataLoader (This will work perfectly now!)
loader = DataLoader( 
    dataset,
    batch_size = 2,
    shuffle = False # here shuffle is not done but shuffling is done to make
                    # the dataset better like mixed.
)

print("\nBatches:")

# 3. Print the batches
for batch in loader:
    print(batch)
```
