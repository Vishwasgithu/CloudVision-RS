```python
x = torch.tensor([1,2,3])
print(x.shape)  # Tensor shape is torch.Size([3]) which means it is a 1-dimensional tensor (a vector) with 3 elements. The shape of a tensor is represented as a tuple of integers, where each integer represents the size of the corresponding dimension. In this case, the tensor has only one dimension with a size of 3.
[3] means it is a 1-dimensional tensor (a vector) with 3 elements. The shape of a tensor is represented as a tuple of integers, where each integer represents the size of the corresponding dimension. In this case, the tensor has only one dimension with a size of 3.
```

```python
x = torch.tensor( [ [1, 2, 3] ,
                   [4, 5 ,6]
])
print(x)
print(x.shape) # Tensor shape is torch.Size([2, 3]) which means it is a 2-dimensional tensor (a matrix) with 2 rows and 3 columns. The shape of a tensor is represented as a tuple of integers, where each integer represents the size of the corresponding dimension. In this case, the tensor has two dimensions: the first dimension has a size of 2 (indicating the number of rows), and the second dimension has a size of 3 (indicating the number of columns).
```

```python
device = "cuda"
x_gpu = torch.tensor([1,2,3]).to(device)
print(x_gpu)
print(x_gpu.device)
```

```python
# Shifting the tensor to the GPU for blazing-fast speed
x_gpu = x.to('cuda') # Shortest methos to shift a tensor to the gpu is to use the .to() method and specify the device you want to shift it to. In this case, we are shifting the tensor x to the GPU by passing 'cuda' as an argument to the .to() method. This will move the tensor from the CPU memory to the GPU memory, allowing for faster computations when performing operations on the tensor.
print(x_gpu)  
```
